Write a program to take 3 sides lengths and prints the triangle is equilateral , isoceleas or scanlene

```python
a = int(input())
b = int(input())
c = int(input())

if a == b == c:
  print("Equilateral triangle")
elif a == b or b == c or c == a:
  print("Isosceles triangle")
else:
  print("Scalene triangle")
```

Write a program that prints "Fizz" if a number is divisible by 3, "Buzz" if divisible by 5, "FizzBuzz" if divisible by both, or the number itself otherwise.

```python
a = int(input())

if a % 3 == 0 and a % 5 == 0:
  print("FizzBuzz")
elif a % 3 == 0:
  print("Fizz")
elif a % 5 == 0:
  print("Buzz")
```

write a program to check if a given string is a palindrome(reads the same backward)

```python
a = input("Enter the String : ")
if a == a[::-1]:
  print("Palindrome")
else:
  print("Not Palindrome")
```

**while loop**

```python
n = int(input())

sum = 0
i = 2

while i <=n:
  sum += i
  i += 2
  ## 2 + 4 + 6 + 8 + 10 + 12 + 14+ 16
  print(sum)
```

Nested while loop

```python
outer_count = 0
while outer_count < 3:
  print("Outer loop iteration" ,outer_count)

  inner_count = 0
  while inner_count < 2:
    print("Inner loop iteration", inner_count)
    inner_count += 1

  outer_count += 1
```

Find the sunm of all even numbers from 1 to 100 using a for loop.

Write a program to calculate the factorial of a number using a while loop

```python
num = int(input("Enter a number: "))
fact = 1
while num > 1:
  fact *= num
  num -= 1 #decrement karna h isliye - lagaya hai!!
print(fact)
```
