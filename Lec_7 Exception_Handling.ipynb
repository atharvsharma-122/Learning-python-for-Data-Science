Exception Handling

```python
try:
  ## critical code
  num = int(input("Enter a number: "))
  result = 10/ num
except ZeroDivisionError:
    print("Cannot divide by 0")
    ## handle the errors/ exception
else:
  ## will be excecuted only if try block will run properly
  print(result)
finally:
  ## will be excecuted indendent of wheather try or except block will run!!7
  ## typically used to clean up actions, like closing files or releasing resources
  print("This is always executed")
```
