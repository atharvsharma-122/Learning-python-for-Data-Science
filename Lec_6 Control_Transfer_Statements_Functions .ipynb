Create a function "calc" in which it is having add functionality, subtraction functionality, multiplication and division functionality of two numbers.

```python
def calc(val1, val2, operation = 'add'):
  if operation == 'add':
    return val1 + val2
  elif operation == 'sub':
    return val1 - val2
  elif operation == "mul":
    return val1 * val2
  elif operation == "div":
    return val1 / val2
result =calc(10, 20 , "mul")
print(result)
```

Armstrong Number

```python
## function
def is_armstrong(num):
  str_num = str(num)
  num_digits = len(str_num)

  sum_of_powers = sum(int(digit)**num_digits for digit in str_num)
  return sum_of_powers == num



##function calling
result = is_armstrong(153)
print(result)
```

```python
def sum_and_product(num1, num2):
  sum_result = num1 + num2
  product_result = num1 * num2
  return sum_result , product_result

num1 = 6
num2 = 10
sum_result , product_result = sum_and_product(num1, num2)
print("Sum:" , sum_result)
print("Product:" , product_result)
```
